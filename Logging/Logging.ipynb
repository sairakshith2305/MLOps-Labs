```python
import logging
```

```python
for handler in logging.root.handlers[:]:
    logging.root.removeHandler(handler)
```

```python
logging.basicConfig(
    level=logging.DEBUG,
    format='%(asctime)s - %(levelname)s - %(message)s',
    force=True
)
```

```python
order_logger = logging.getLogger("FoodDelivery")

menu = {
    "pizza": 12.99,
    "burger": 8.99,
    "pasta": 10.99,
    "salad": 6.99
}

cart = []
customer_balance = 50.00
```

```python
def browse_menu():
    """Customer browses the menu"""
    order_logger.debug("Customer opened menu page")

    print("\nAvailable Items:")
    for item, price in menu.items():
        print(f"  {item.title()}: ${price}")

    order_logger.info(f"Menu displayed with {len(menu)} items")

browse_menu()
```

```python
def add_to_cart(item):
    """Add item to shopping cart"""
    order_logger.debug(f"Attempting to add '{item}' to cart")

    if item not in menu:
        order_logger.error(f"Item not found in menu: '{item}'")
        print(f"Sorry, '{item}' is not available")
        return False

    cart.append(item)
    order_logger.info(f"Added {item} to cart (Cart size: {len(cart)})")
    print(f"{item.title()} added to cart")

    if len(cart) > 5:
        order_logger.warning(f"Large order detected: {len(cart)} items in cart")
        print("That's a big order!")

    return True

print("Adding items to cart")
add_to_cart("pizza")
add_to_cart("burger")
add_to_cart("pasta")
print(f"\n Cart: {cart}")
```

```python
print("Trying to add item not on menu")
add_to_cart("sushi")
```

```python
print("Adding more items to trigger WARNING")
add_to_cart("salad")
add_to_cart("pizza")
add_to_cart("burger")

print(f"\n Cart now has {len(cart)} items")
```

```python
def calculate_total():
    """Calculate order total"""
    total = sum(menu[item] for item in cart)
    order_logger.debug(f"Calculated total: ${total:.2f} for {len(cart)} items")
    return total

def place_order():
    """Process the order"""
    global customer_balance

    if not cart:
        order_logger.error("Attempted to place empty order")
        print("Your cart is empty!")
        return False

    total = calculate_total()
    order_logger.info(f"Processing order: {len(cart)} items, Total: ${total:.2f}")

    if total > customer_balance:
        order_logger.error(f"Payment failed: Insufficient funds (Required: ${total:.2f}, Available: ${customer_balance:.2f})")
        print(f"Insufficient funds! Need ${total:.2f}, have ${customer_balance:.2f}")
        return False

    if total > 100:
        order_logger.critical(f"FRAUD ALERT: Unusually large order: ${total:.2f}")
        print("FRAUD ALERT - Order flagged!")
        return False

    customer_balance -= total
    order_logger.info(f"Payment successful! Charged ${total:.2f}, Remaining: ${customer_balance:.2f}")
    print(f"Order placed! Total: ${total:.2f}")
    print(f"Remaining balance: ${customer_balance:.2f}")

    cart.clear()
    return True
```

```python
print(" Placing your order")
place_order()
```

```python
print("Adding expensive items...\n")
add_to_cart("pizza")
add_to_cart("pizza")
add_to_cart("pasta")
add_to_cart("pasta")

print(f"\nYour balance: ${customer_balance:.2f}")
print("Trying to place order\n")
place_order()
```

```python
def apply_discount(code):
    """Apply discount code"""
    try:
        order_logger.debug(f"Applying discount code: {code}")

        discount_percent = int(code.replace("SAVE", ""))

        if discount_percent <= 0:
            raise ValueError("Invalid discount")

        order_logger.info(f"Discount applied: {discount_percent}% off")
        print(f"{discount_percent}% discount applied!")
        return True

    except ValueError:
        order_logger.exception(f"Failed to apply discount: '{code}'")
        print(f"Invalid discount code: '{code}'")
        return False

print("Trying discount codes")
apply_discount("SAVE20")
print()
apply_discount("INVALID")
```

```python
logging.basicConfig(
    filename='food_orders.log',
    level=logging.DEBUG,
    format='%(asctime)s - %(levelname)s - %(message)s',
    force=True
)

order_logger.info("File logging enabled")
```

```python
def toggle_maintenance_mode(enable):
    """Change logging level"""
    if enable:
        order_logger.setLevel(logging.WARNING)
        order_logger.warning("Maintenance mode ON - only showing WARNING+")
        print("Maintenance mode enabled")
    else:
        order_logger.setLevel(logging.DEBUG)
        order_logger.info("Maintenance mode OFF - showing all logs")
        print("Normal mode restored")

print("Testing log level control")
toggle_maintenance_mode(True)
print()
browse_menu()
print()
toggle_maintenance_mode(False)
```

```python
order_logger = logging.getLogger("FoodDelivery")
order_logger.setLevel(logging.DEBUG)
order_logger.handlers.clear()


console_handler = logging.StreamHandler()
console_handler.setLevel(logging.INFO)

file_handler = logging.FileHandler('food_orders_detailed.log')
file_handler.setLevel(logging.DEBUG)

formatter = logging.Formatter('%(asctime)s - %(name)s - %(levelname)s - %(message)s')
console_handler.setFormatter(formatter)
file_handler.setFormatter(formatter)

order_logger.addHandler(console_handler)
order_logger.addHandler(file_handler)

print(" Multi-handler logging demo:\n")
order_logger.debug("This DEBUG goes to FILE only (not console)")
order_logger.info("This INFO goes to BOTH console and file")
order_logger.warning("This WARNING also goes to BOTH")
```
